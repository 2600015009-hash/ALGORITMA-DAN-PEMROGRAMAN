```python
# menentukan nilai maksimum dari tiga buah bilangan

x = 3
y = 8
z = 10

nilai_maksimum = max(x, y, z)
print(nilai_maksimum)
```

```python
# Menggunakan contoh nilai variabel seperti pada contoh modul

bil = 10
print("Bilangan =", bil)

if bil > 0:
    print("Positif")
elif bil < 0:
    print("Negatif")
elif bil == 0:
    print("Nol")
```
