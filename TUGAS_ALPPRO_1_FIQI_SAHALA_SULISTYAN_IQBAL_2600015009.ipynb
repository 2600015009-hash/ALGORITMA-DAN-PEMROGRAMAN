```python
# Menentukan nilai pi dan jari-jari (r)
pi = 3.14
r = 10  # Contoh nilai jari-jari lingkaran

# Rumus luas lingkaran menggunakan operator pangkat (**)
luas = pi * (r ** 2)

print("Jari-jari lingkaran =", r)
print("Luas lingkaran =", luas)
```

```python
# Daftar harga alat tulis
buku = 3000
pensil = 1000
penghapus = 500

# Persamaan matematika belanjaan Agung (3 buku dan 2 pensil)
total_agung = (3 * buku) + (2 * pensil)

# Persamaan matematika belanjaan Sandy (2 buku dan 4 penghapus)
total_sandy = (2 * buku) + (4 * penghapus)

# Total keseluruhan uang yang dibutuhkan keduanya
total_bersama = total_agung + total_sandy

print("Total uang yang dibutuhkan Agung =", total_agung)
print("Total uang yang dibutuhkan Sandy =", total_sandy)
print("Total uang gabungan Agung dan Sandy =", total_bersama)
```
